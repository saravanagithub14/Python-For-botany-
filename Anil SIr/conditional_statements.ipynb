```python
# if a number is divisible by 5 , print fizz, if a number is divisible
# by 7 print buzz, if a number is divisible by 5 and 7 print fizz buzz
num = int(input('enter a number'))
if num%5 == 0 and num%7 ==0:
    print("fizzbuzz")
elif num%5 == 0:
    print('FIZZ')
elif num%7 ==0:
    print('buzz')
```
